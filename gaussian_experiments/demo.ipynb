```python
import numpy as np
import matplotlib.pyplot as plt

def double_peak_function(x, peak1_center, peak1_width, peak2_center, peak2_width, amp1=1, amp2=1, noise_level=0.):
    """
    A function with two peaks, represented as a sum of two Gaussian functions.

    Parameters:
    x: numpy array, the input range.
    peak1_center: float, center of the first peak.
    peak1_width: float, width of the first peak.
    peak2_center: float, center of the second peak.
    peak2_width: float, width of the second peak.
    amp1: float, amplitude of the first peak.
    amp2: float, amplitude of the second peak.

    Returns:
    numpy array, the values of the function.
    """
    peak1 = amp1 * np.exp(-((x - peak1_center) ** 2) / (2 * peak1_width ** 2))
    peak2 = amp2 * np.exp(-((x - peak2_center) ** 2) / (2 * peak2_width ** 2))
    noise = np.random.normal(0, noise_level, x.shape)
    return peak1 + peak2 + noise

# Define the range and parameters for the peaks
x = np.linspace(-10, 10, 500)
peak1_center = -3
peak1_width = 1
peak2_center = 4
peak2_width = 2
amp1 = 1.5
amp2 = 1

for noise_level in [0.5, 0.4, 0.3, 0.2, 0.1, 0.0]:
    # Compute the function values
    y = double_peak_function(x, peak1_center, peak1_width, peak2_center, peak2_width, amp1, amp2, noise_level)
    
    # Plot the function
    fig, ax = plt.subplots()
    
    ax.scatter(x, y, label="Double Peak Function", color="blue")
    
    # Hide all tick labels
    ax.set_xticks([])
    ax.set_yticks([])
    
    # Add arrowed axes
    ax.spines['left'].set_position('zero')
    ax.spines['bottom'].set_position('zero')
    ax.spines['left'].set_color('black')
    ax.spines['bottom'].set_color('black')
    
    ax.spines['right'].set_color('none')
    ax.spines['top'].set_color('none')
    
    # Add arrows to axes
    arrow_params = dict(arrowstyle='->', color='black', linewidth=1.5)
    # ax.annotate('', xy=(10, 0), xytext=(0, 0), arrowprops=arrow_params)  # X-axis arrow
    # ax.annotate('', xy=(0, 2), xytext=(0, 0), arrowprops=arrow_params)  # Y-axis arrow
    
    # Set aspect ratio and limits
    ax.set_xlim(-10, 10)
    ax.set_ylim(-0.5, 2.5)
    
    plt.show()
# plt.savefig('q.svg',transparent=True)
```
