```python
import sys
import matplotlib.pyplot as plt
sys.path.append('/home/haitong/PycharmProjects/diffusion-model-hallucination/gaussian_experiments')
from ddpm_torch.diffusion import get_beta_schedule, GaussianDiffusion
```

```python
fig, ax = plt.subplots(4, 4, figsize=(10, 8))
for i, beta_schedule in enumerate(['linear', 'warmup10', 'cosine', 'jsd']):
    betas = get_beta_schedule(beta_schedule, 0.0001, 1.0, 20)
    ax[i][0].plot(betas)
    ax[i][0].set_title("beta")
    diffusion = GaussianDiffusion(betas, 'mean', 'fixed-large', 'MSE')
    sqrt_bar_alpha = diffusion.sqrt_alphas_bar
    ax[i][1].plot(sqrt_bar_alpha)
    ax[i][1].set_title("$\\sqrt{\\bar{\\alpha}}$")
    sqrt_recip_alphas_bar = diffusion.sqrt_recip_alphas_bar
    ax[i][2].plot(sqrt_recip_alphas_bar)
    ax[i][2].set_title("$\\sqrt{\\frac{1}{\\bar{\\alpha}}}$")
    sqrt_recip_m1_alphas_bar = diffusion.sqrt_recip_m1_alphas_bar
    ax[i][3].plot(sqrt_recip_m1_alphas_bar)
    ax[i][3].set_title("$\\sqrt{\\frac{1}{\\bar{\\alpha}} - 1}$")

plt.tight_layout()
```
